# Exemplo de análise

Este notebook apresenta um fluxo simples de exploração de dados usando pandas e matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Bibliotecas carregadas com sucesso!')
```

```python
# Ajuste do caminho conforme a estrutura do projeto
df = pd.read_csv('../data/raw/breast_cancer_wisconsin.csv')

df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
# Exemplo de contagem por classe
# Usa a coluna correta da variável alvo (diagnosis) em vez de depender da última coluna do DataFrame
if 'diagnosis' not in df.columns:
    raise ValueError("A coluna 'diagnosis' não foi encontrada no dataset.")

contagem = df['diagnosis'].value_counts().sort_index()
print(contagem)

plt.figure(figsize=(6, 4))
contagem.plot(kind='bar', color=['steelblue', 'salmon'], edgecolor='black')
plt.title('Distribuição da variável alvo')
plt.xlabel('Classe')
plt.ylabel('Quantidade')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
